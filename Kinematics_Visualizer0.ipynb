```python

import matplotlib.pyplot as plt

times = []
positions = []

for t in range(9):
    position = u * t + 0.5 * a * t**2

    times.append(t)
    positions.append(position)

plt.plot(times, positions, marker="o")

plt.xlabel("Time (seconds)")
plt.ylabel("Position (meters)")
plt.title("Position vs Time")

plt.grid()
plt.show()
```

```python

velocities = []

for t in range(9):
    velocity = u + a * t
    velocities.append(velocity)

plt.plot(times, velocities, marker="o")

plt.xlabel("Time (seconds)")
plt.ylabel("Velocity (m/s)")
plt.title("Velocity vs Time")

plt.grid()
plt.show()
```

```python

print("KINEMATICS PROBLEM SOLVER")
print("-------------------------")

# Given information
distance_5th_second = 7
last_second_fraction = 1 / 64

# Calculate total time
T = (1 / last_second_fraction) ** 0.5

# Calculate acceleration
a = distance_5th_second / (4.5 - T)

# Calculate initial velocity
u = -a * T

# Calculate total distance
total_distance = u * T + 0.5 * a * T**2

print("\nRESULTS")
print("Initial velocity =", u, "m/s")
print("Acceleration =", a, "m/s^2")
print("Total time =", T, "seconds")
print("Total distance =", total_distance, "m")
```
